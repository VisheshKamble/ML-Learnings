```python
import numpy as np 
arr = np.array([2,3,4,6,7,8,9,12,13,16,17,23,25,27,34,37,201])
arr
```

```python
# Find the Q1 and Q3 of the array
Q1 = np.percentile(arr , 25)
Q3 = np.percentile(arr , 75)
print("Q1: ", Q1)
print("Q3: ", Q3)
```

```python
# Finding the IQR
IQR = Q3 - Q1
print("IQR: ", IQR)
```

```python
#FInding the lower and upper bound
UF = Q3 + 1.5*IQR
LF = Q1 - 1.5*IQR
print("Upper bound: ", UF)
print("Lower bound: ", LF)
```

```python
#Removing the outliers
l = []
for i in arr:
    if i >= LF and i <= UF:
        l.append(i)
print("Outliers: ", l)
```
