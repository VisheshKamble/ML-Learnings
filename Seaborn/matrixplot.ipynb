**Heat Maps**

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
flight = sns.load_dataset('flights')
flight
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
tipscorr = tips[['total_bill', 'tip', 'size']].corr()
tipscorr
```

```python
sns.heatmap(tipscorr, annot = True)
```

```python
flightscorr = flight[['passengers', 'year']].corr()
flightscorr
```

```python
sns.heatmap(flightscorr, annot = True) 
```

**Cluster Map**

```python
sns.clustermap(tipscorr, annot = True)
```

**Pivot Table Map**

```python

pivot_flights = flight.pivot_table(index = 'month', columns = 'year', values = 'passengers')
pivot_flights
```

```python
sns.heatmap(pivot_flights)
```
