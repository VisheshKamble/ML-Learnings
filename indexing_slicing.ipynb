```python
arr = np.arange(1,31).reshape(5,6)
arr
```

```python
arr[4,4]
```

```python
arr[2:5,4:6]
```

BOOLEAN INDEXING 

```python
arr = np.arange(1,21)
arr
```

```python
boolean_arr = arr % 2 == 0
boolean_arr
```

```python
arr = arr[boolean_arr]
arr
```
