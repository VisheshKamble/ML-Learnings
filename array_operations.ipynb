```python
import numpy as np 
```

Arithmetic Operations

```python
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])

arr1
arr2
```

```python
arr_sum = arr1 + arr2
arr_sum 
```

```python
arr_sub  = arr1 - arr2
arr_sub
```

```python
arr_mul = arr1 * arr2
arr_mul
```

```python
arr_mod = arr1 % arr2
arr_mod
```

```python
arr_div = arr1 / arr2
arr_div
```

Broadcasting

```python
l = [1,2,3,4,5]
arr = np.array(l)
arr
```

```python
arr + 10 
```

```python
arr2 = np.arange(1,26).reshape(5,5)
arr2
```
