```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

**Basic Plot Using Matplotlib**

```python
x = np.linspace(0, 5, 11)
x
y = x ** 2
y
```

```python
plt.plot(x, y)
plt.title("Matplotlib Basic Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
```

**SubPlots in Matplotlib**

```python
plt.subplot(2,2,1)
plt.plot(x , y)
plt.subplot(2,2,2)
plt.plot(y , x)
plt.subplot(2,2,3)
plt.plot(x , y**3)
plt.subplot(2,2,4)
plt.plot(x , y**10)
```

**Object Oriented Way**

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x , y)
axis2 = fig.add_axes([0.2 , 0.5 , 0.2, 0.2])
axis2.plot(y , x)
```
